```sql
%python
df_coverage_provider = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(f"{RAW_DATA_PATH}/coverage_provider.csv")
)

df_coverage_provider.printSchema()

qtd_rows = df_coverage_provider.count()
print(f"Quantidade de registros em coverage_provider: {qtd_rows:,}")
df_coverage_provider.limit(100).display()
```

```sql
%python
(
    df_coverage_provider.write
    .mode("overwrite")
    .saveAsTable("dataquality_challenge.bronze.coverage_provider")
)

print("Tabela dataquality_challenge.bronze.coverage_provider criada com sucesso!")
```
